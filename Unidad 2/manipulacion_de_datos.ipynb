## U2.03  - Transformaciones de columnas

###	Columnas calculadas con operaciones númericas

```python
import numpy as np #librería para operaciones numéricas

df =pd.read_csv('ventas_completas.csv')
print(list(df.columns))
```

```python
df['precio_original'] = df['total'] / df['cantidad']
df['total_con_iva'] = df['total'] * 1.19
df['log_total'] = np.log(df['total'])
df['sqrt_total'] = np.sqrt(df['total'])
df[['total', 'cantidad', 'precio_original','total_con_iva','log_total','sqrt_total']].round(2)   
```

### Columnas Calculadas con operaciones de texto.

Operaciones con Columnas de Texto

```python
df['marca']= df['producto'].str.split(' ').str[1]
df['es_laptop'] = df['producto'].str.contains('laptop', case=False)
df['largo_nombre'] = df['producto'].str.len()
df['producto_largo'] = df['producto'] + '(' + df['largo_nombre'].astype(str) + ' caracteres)'
df[['producto', 'marca', 'es_laptop', 'largo_nombre', 'producto_largo']]
```

### Columnas calculada con condicionales lógicas

```python
df['tipo_venta'] = np.where(df['total']>1000, 'Alta', 'Baja')

condiciones = [
    (df['total']>=2000),
    (df['total']>=1000),
    (df['total']>=500)
]
valores = ['Premium', 'Alta', 'Media']

df['categoria_venta'] = np.select(condiciones, valores, default='Baja')
df[['total', 'tipo_venta', 'categoria_venta']]
```

## U2.04  - Agrupaciones y agregaciones

###  Agrupar datos con .groupby()

```python
import pandas as pd
import numpy as np #librería para operaciones numéricas
df =pd.read_csv('ventas_completas.csv')
print(list(df.columns))
```

```python
df.groupby(['region', 'estado'])['total'].sum().sort_values(ascending=False).reset_index()
#df.groupby(['region', 'estado'])['total'].sum().reset_index().sort_values(by='total', ascending=False)
```

```python
df.groupby('region')['total'].sum()
```

### Funciones de agregación: .sum(), .mean(), .count() y agg()

```python
df.groupby(['region', 'estado'])['total'].agg(['sum', 'mean', 'max', 'min']).sort_values(by='sum', ascending=False).reset_index().round(2)
```

```python
df.groupby(['region', 'estado']).agg({
    'total':['sum', 'mean', 'max', 'min'], 
    'cantidad':['sum','mean']
    }).reset_index().round(2).sort_values(by=('total','sum'), ascending=False)
```

### Análisis por categorías y segmentos

```python
pd.crosstab(
    df['region'],
    df['estado'],
    values=df['total'],
    aggfunc='sum'
).round(2)
```

```python
df.pivot_table(
    values='total',
    index=['region', 'estado'],
    columns='categoria',
    aggfunc='sum',
    fill_value=0
).reset_index().round(2)
```

## U2.05 - Uniones y combinaciones 

Diferencias clave entre merge y concat:

- **merge():** Une DataFrames por columnas comunes

- **concat():** Apila DataFrames vertical u horizontalmente


### Combinaciones de tablas con Merge.

Resumen de tipos de join:

| Tipo  | Descripción                                  | Cuándo usar                            |
| ----- | -------------------------------------------- | -------------------------------------- |
| inner | Solo registros que coinciden en ambas tablas | Necesitas datos completos              |
| left  | Todos los registros de la tabla izquierda    | Base principal + información adicional |
| right | Todos los registros de la tabla derecha      | Menos común, similar a left invertido  |
| outer | Todos los registros de ambas tablas          | Análisis completo, detectar faltantes  |

Tipos de Join

```python
import pandas as pd
#Crear datasets de ejemplo
productos_ejemplo = pd.DataFrame(
    {
    'id_producto': [1, 2, 3, 4],
    'productos  ': ['Laptop', 'Tablet', 'Smartphone', 'Monitor'],
    'monto_unit': [100, 200, 150, 300]
    }
)

ventas_ejemplo = pd.DataFrame(
    {
    'id_venta': [1, 2, 3, 4, 5, 6],
    'cliente': ['Ana', 'Ana', 'María', 'Juan', 'José', 'Miguel'],
    'id_producto': [1, 2, 5, 6, 1, 1],
    'cantidad ': [2, 1, 3, 2, 1, 1]
    }
)

pd.merge(
ventas_ejemplo, 
productos_ejemplo, 
how='inner',
on='id_producto'
)
```

### Uniones de tablas con Concat

```python
#Crear dataframe de ejemplo: ventas de enero
df_enero = pd.DataFrame(
    {
    'fecha': ['2024-01-05', '2024-01-12', '2024-01-18', '2024-01-25'],
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor'],
    'cantidad': [2, 5, 3, 1],
    'total': [2400, 125, 255, 350]
    }
)
#Crear dataframe de ejemplo: ventas de enero
df_febrero = pd.DataFrame(
    {
    'fecha': ['2024-02-03', '2024-02-14', '2024-02-22'],
    'producto': ['Laptop', 'Auriculares', 'Mouse'],
    'cantidad': [1, 2, 3],
    'total': [1200, 130, 75]
    }
)
pd.concat([df_enero, df_febrero], ignore_index=True)
```

```python
#Crear dataframes con diferentes columnas pero mismas filas
info_base = pd.DataFrame({
    'id_venta': [1, 2, 3, 4, 5],
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Webcam'],
    'total': [1200, 25, 85, 350, 45]
    })

info_adicional = pd.DataFrame({
    'region': ['Norte', 'Sur', 'Este', 'Oeste', 'Centro'],
    'vendedor': ['Ana García', 'Carlos Méndez', 'María López', 'Juan Pérez', 'Lucía Rodríguez'],
    'estado': ['Completado', 'Pendiente', 'Completado', 'Completado', 'Cancelado']
    }) 

pd.concat([info_base, info_adicional], axis=1)
```
