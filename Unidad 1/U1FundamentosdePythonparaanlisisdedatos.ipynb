## U1.03 - Estructuras de control y funciones

###	Condicionales (if, elif, else)

```python
monto = 900
if monto >= 1500:
    print("Alto")
elif monto >= 1000:
    print("Medio")
else:
    print("Bajo")
```

### Bucles (for, while) para iterar sobre colecciones

```python
# Bucles for para iterar listas

ventas = [1200, 1500, 980, 2100, 1750]
total=0
for x in ventas:
    total += x
    print("Total de ventas es: ",total)

```

```python
# Bucles for para iterar diccionarios

cliente={"nombre": "Juan", "ciudad": "Barcelona", "edad": 35}
for clave, valor in cliente.items():
    print(f'{clave}: {valor}')

```

```python
# Bucles while para condiciones

ventas_diarias = [500, 750, 600, 900, 1200]
objetivo = 2500
acumulado = 0
dia = 0
while acumulado < objetivo and dia < len(ventas_diarias):
    acumulado += ventas_diarias[dia]
    dia += 1
print(f'Objetivo alcanzado en {dia} días con un total de {acumulado} ventas') 

```

### Creación de funciones que reciben y retornan datos

```python
def calcular_descuento(precio, descuento):
    """Calcula el precio final con descuento aplicado"""
    precio_descuento = precio - (precio * descuento / 100)
    precio_sin_descuento = precio
    return precio_descuento, precio_sin_descuento

precio_final = calcular_descuento(2000, 15)
print(f'El precio final con descuento es: ${precio_final[0]}, precio sin descuento: ${precio_final[1]} el ahorro es: ${precio_final[1]-precio_final[0]}')
print(calcular_descuento(2000, 15))
```

### Manejo básico de errores con try/except

```python
def calcular_margen(precio_venta, precio_costo):
    """Calcula el margen de ganancia"""
    try:
        margen = (precio_venta - precio_costo) / precio_costo * 100
        return round(margen, 2)
    except ZeroDivisionError:
        return "Error división por cero"
    except TypeError:
        return "Error: Tipo de dato no válido"
    except:
        return "Error desconocido"

#pruebas
print(calcular_margen(150, 100)) # 50.0
print(calcular_margen(150, 0)) # Error: Costo no puede ser cero
print(calcular_margen("150", 100)) # Error: Tipo de dato no válido
```
